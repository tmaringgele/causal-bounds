```python
%%manim -v WARNING -qm CausalProbDerivation
from manim import *

class CausalProbDerivation(Scene):
    def construct(self):
        # Step 1: Write P(X=0, Y=0)
        step1 = MathTex(r"P(X=0, Y=0)")
        self.play(Write(step1))
        self.wait(1)

        # Step 2: Add subscript 0 to Y => Y_0
        step2 = MathTex(r"P(X=0, Y=0) = P(X=0, Y_0=0)")
        self.play(Transform(step1, step2))
        self.wait(1)

        # Step 3: Use law of total probability
        step3 = MathTex(
            r"P(X=0, Y=0) = \sum_{y_1} P(X=0, Y_0=0, Y_1=y_1)"
        )
        self.play(Transform(step1, step3))
        self.wait(1)

        # Step 4: Expand the sum explicitly
        step4 = MathTex(
            r"P(X=0, Y=0) = P(X=0, Y_0=0, Y_1=0) + P(X=0, Y_0=0, Y_1=1)"
        )
        self.play(Transform(step1, step4))
        self.wait(1)

        # Step 5: Switch to short notation q_0 and q_2
        step5 = MathTex(r"P(X=0, Y=0) = q_0 + q_2")
        self.play(Transform(step1, step5))
        self.wait(1)

        # Step 6: Plug in values
        step6 = MathTex(r"0.3 = q_0 + q_2")
        self.play(Transform(step1, step6))
        self.wait(2)
```

```python
%%manim -v WARNING -qm  CausalProbDerivationPartialTransform 

from manim import *

config.frame_width = 20
config.frame_height = 6


class CausalProbDerivationPartialTransform(Scene):
    def construct(self):
    
        # Initial LHS and first RHS
        lhs = MathTex(r"P(X=0, Y=0)")
        eq = MathTex("=")
        rhs1 = MathTex(r"P(X=0, Y_0=0)")

        group = VGroup(lhs, eq, rhs1).arrange(RIGHT, buff=0.3)
        group.to_edge(LEFT, buff=1) 
        self.play(Write(group))
        self.wait(1)

        # Step 3: Add sum expression on RHS
        rhs2 = MathTex(r"\sum_{y_1} P(X=0, Y_0=0, Y_1=y_1)")
        rhs2.move_to(rhs1, aligned_edge=LEFT)
        self.play(Transform(rhs1, rhs2))
        self.wait(1)

        # Step 4: Expand the sum explicitly
        rhs3 = MathTex(
            r"P(X=0, Y_0=0, Y_1=0) + P(X=0, Y_0=0, Y_1=1)"
        )
        rhs3.move_to(rhs1, aligned_edge=LEFT)
        self.play(Transform(rhs1, rhs3))
        self.wait(1)

        # Step 5: Use notation q_0 + q_2
        rhs4 = MathTex(r"q_0 + q_2")
        rhs4.move_to(rhs1, aligned_edge=LEFT)
        self.play(Transform(rhs1, rhs4))
        self.wait(1)

        # Step 6: Plug in actual value
        final_eq = MathTex(r"0.3 = q_0 + q_2")
        final_eq.move_to(group)
        self.play(Transform(group, final_eq))
        self.wait(2)
```

```python
%%manim -v WARNING -qm  CausalProbDerivationPartialTransform 

from manim import *

config.frame_width = 20
config.frame_height = 6


class CausalProbDerivationPartialTransform(Scene):
    def construct(self):

        # Pre-create all equation parts so position doesn't shift
        lhs = MathTex(r"P(X=0, Y=0)")
        eq = MathTex("=")
        rhs1 = MathTex(r"P(Y=0 \mid X=0)\,P(X=0)")

        group = VGroup(lhs, eq, rhs1).arrange(RIGHT, buff=0.3)
        group.to_edge(LEFT, buff=1)

        # Step 0: Show LHS first
        self.play(Write(lhs))
        self.wait(1)

        # Step 1: Show equals sign and first RHS
        self.play(Write(eq), Write(rhs1))
        self.wait(1)

        # Step 2: Consistency assumption
        rhs2 = MathTex(r"P(Y_0=0 \mid X=0)\,P(X=0)").move_to(rhs1, aligned_edge=LEFT)
        self.play(Transform(rhs1, rhs2))
        self.wait(1)

        # Step 2.5: Reverse chain rule to joint
        rhs3 = MathTex(r"P(Y_0=0, X=0)").move_to(rhs1, aligned_edge=LEFT)
        self.play(Transform(rhs1, rhs3))
        self.wait(1)

        # Step 3: Law of Total Probability over Y_1
        rhs4 = MathTex(r"\sum_{y_1} P(Y_1=y_1, Y_0=0, X=0)").move_to(rhs1, aligned_edge=LEFT)
        self.play(Transform(rhs1, rhs4))
        self.wait(1)

        # Step 4: Expand the sum explicitly
        rhs5 = MathTex(
            r"P(Y_1=0, Y_0=0, X=0) + P(Y_1=1, Y_0=0, X=0)"
        ).move_to(rhs1, aligned_edge=LEFT)
        self.play(Transform(rhs1, rhs5))
        self.wait(1)

        # Step 5: Insert decision variables
        rhs6 = MathTex(r"q_0 + q_4").move_to(rhs1, aligned_edge=LEFT)
        self.play(Transform(rhs1, rhs6))
        self.wait(1)

        # Step 6: Insert value
        final_eq = MathTex(r"0.3 = q_0 + q_4").move_to(group, aligned_edge=LEFT)
        self.play(Transform(group, final_eq))
        self.wait(2)
```
